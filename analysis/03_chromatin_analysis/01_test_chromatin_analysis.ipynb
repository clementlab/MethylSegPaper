# Chromatin Analysis Migration Smoke Test


```python
from pathlib import Path
import sys

REPO_ROOT = Path.cwd()
while not (REPO_ROOT / "repo_paths.py").exists():
    if REPO_ROOT.parent == REPO_ROOT:
        raise RuntimeError("Could not locate repo root from notebook cwd")
    REPO_ROOT = REPO_ROOT.parent
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

chromatin_dir = REPO_ROOT / "analysis" / "03_chromatin_analysis"
if str(chromatin_dir) not in sys.path:
    sys.path.insert(0, str(chromatin_dir))

from repo_paths import CHROMATIN_DATA_DIR, CHROMATIN_RESULTS_DIR, REGION_CALLING_RESULTS_DIR
from chromatin_analysis_utils import load_peak_regions, resolve_chromatin_paths
```

```python
sample = "TE5.wgbs"
bw_path, peak_path = resolve_chromatin_paths(CHROMATIN_DATA_DIR, sample)
smoke_dir = CHROMATIN_RESULTS_DIR / "smoke_test"
smoke_dir.mkdir(parents=True, exist_ok=True)
fixture_path = smoke_dir / "tiny_peaks.bed.gz"
import gzip
with gzip.open(fixture_path, "wt") as handle:
    handle.write("chr1\t100\t200\nchr1\t220\t260\n")
peak_df = load_peak_regions(fixture_path)
assert not peak_df.empty, fixture_path
print({
    "sample": sample,
    "resolved_bw_path": str(bw_path),
    "resolved_peak_path": str(peak_path),
    "fixture_peak_rows": int(len(peak_df)),
    "segmentation_root": str(REGION_CALLING_RESULTS_DIR),
})
```
