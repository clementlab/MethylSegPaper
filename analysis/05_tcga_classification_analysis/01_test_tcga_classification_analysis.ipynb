```python
from pathlib import Path
import sys

REPO_ROOT = Path.cwd()
while not (REPO_ROOT / "repo_paths.py").exists():
    if REPO_ROOT.parent == REPO_ROOT:
        raise RuntimeError("Could not locate repo root from notebook cwd")
    REPO_ROOT = REPO_ROOT.parent
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

for import_path in [
    REPO_ROOT / "analysis" / "05_tcga_classification_analysis",
    REPO_ROOT / "analysis" / "05_tcga_classification_analysis" / "utils",
]:
    import_str = str(import_path)
    if import_str not in sys.path:
        sys.path.insert(0, import_str)

from repo_paths import METHYLATION_DATA_DIR, REFERENCE_DATA_DIR, TCGA_CLASSIFICATION_RESULTS_DIR
from tcga_segmentation_workflow import load_tcga_sample_beta_dataframe
```
