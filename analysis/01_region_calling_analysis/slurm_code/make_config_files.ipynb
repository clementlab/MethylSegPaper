```python
from pathlib import Path
import yaml

config_files = Path("configs/")
config_files.mkdir(parents=True, exist_ok=True)
data_path = Path(
    "/uufs/chpc.utah.edu/common/home/u0914269/clement/projects/20260624_methylseg/data/methylation_data"
)
configs = []
for beta_file in data_path.glob("*.beta"):
    sample_id = beta_file.stem
    if sample_id.endswith(".hg38"):
        continue
    config_details = {
        "sample": sample_id,
        "genome": "hg19",
        "meth_file": str(beta_file.absolute()),
    }
    file_path = config_files / f"{sample_id}_config.yaml"
    configs.append(file_path)

    with open(file_path, "w") as file_handle:
        yaml.dump(config_details, file_handle, default_flow_style=False)

for bed_file in data_path.glob("*.bed.gz"):
    sample_id = Path(bed_file.stem).stem
    config_details = {
        "sample": sample_id,
        "genome": "hg38",
        "meth_file": str(bed_file.absolute()),
    }
    file_path = config_files / f"{sample_id}_config.yaml"
    configs.append(file_path)

    with open(file_path, "w") as file_handle:
        yaml.dump(config_details, file_handle, default_flow_style=False)
```

```python
count = 0
with open("configs.txt", "w") as configs_file:
    for f in configs:
        configs_file.write(str(f.absolute()) + "\n")
        count += 1
        if count % 500 == 0:
            break
```
