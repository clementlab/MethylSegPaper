# LAD Analysis Migration Smoke Test


```python
from pathlib import Path
import sys

REPO_ROOT = Path.cwd()
while not (REPO_ROOT / "repo_paths.py").exists():
    if REPO_ROOT.parent == REPO_ROOT:
        raise RuntimeError("Could not locate repo root from notebook cwd")
    REPO_ROOT = REPO_ROOT.parent
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

lad_dir = REPO_ROOT / "analysis" / "04_lad_analysis"
if str(lad_dir) not in sys.path:
    sys.path.insert(0, str(lad_dir))

from repo_paths import LAD_RESULTS_DIR, REFERENCE_DATA_DIR, REGION_CALLING_RESULTS_DIR
from lad_analysis_utils import read_bed_intervals, read_bedgraph_signal, read_chrom_sizes
```

```python
lad_path = REFERENCE_DATA_DIR / "LAD_intervals.bed"
signal_path = REFERENCE_DATA_DIR / "laminB1_signal.bedGraph"
chrom_sizes_path = REFERENCE_DATA_DIR / "hg38.chrom.sizes"
lad_df = read_bed_intervals(lad_path)
signal_df = read_bedgraph_signal(signal_path)
chrom_sizes = read_chrom_sizes(chrom_sizes_path)
assert not lad_df.empty, lad_path
assert not signal_df.empty, signal_path
assert chrom_sizes["chr1"] > 0
smoke_dir = LAD_RESULTS_DIR / "smoke_test"
smoke_dir.mkdir(parents=True, exist_ok=True)
print({
    "lad_intervals": int(len(lad_df)),
    "signal_rows": int(len(signal_df)),
    "chrom_sizes": len(chrom_sizes),
    "segmentation_root": str(REGION_CALLING_RESULTS_DIR),
    "smoke_dir": str(smoke_dir),
})
```
