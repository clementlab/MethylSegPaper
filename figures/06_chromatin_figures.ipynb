# Chromatin Figures

Regenerate chromatin `plotProfile` and `plotHeatmap` outputs from the per-tool deepTools matrices written by `analysis/03_chromatin_analysis/run_chromatin.py`.


```python
%load_ext autoreload
%autoreload 2

import shutil
import subprocess
from pathlib import Path

import pandas as pd
from IPython.display import Image, Markdown, display

from utils.figures_utils import CHROMATIN_FIGURE_OUTPUT_DIR, RESULTS_DIR
```

```python
from colors import CHROMATIN_PROFILE_COLORS, TOOL_HIGHLIGHT_COLORS

OUT_DIR = CHROMATIN_FIGURE_OUTPUT_DIR
OUT_DIR.mkdir(parents=True, exist_ok=True)

SAVE_FIGURES = True
SELECTED_SAMPLES = None
SELECTED_GROUPS = None
SELECTED_SOURCE_IDS = None

CHROMATIN_RESULTS_DIR = RESULTS_DIR / "03_chromatin_analysis"
DEEPTOOLS_OUTPUTS_PATH = CHROMATIN_RESULTS_DIR / "tables" / "deeptools_outputs.tsv"
DEEPTOOLS_REGION_MANIFEST_PATH = CHROMATIN_RESULTS_DIR / "tables" / "deeptools_region_manifest.tsv"
DISPLAY_GROUP_SPECS = [
    {
        "group_id": "cleaned-wgbs-tools",
        "group_label": "Cleaned WGBS Tools",
        "source_ids": [
            "methylseg__cleaned",
            "methylseekr__source-default",
            "dnmtools__source-default",
            "dnmtools_pmr__source-default",
            "mmseekr__source-default",
            "methyl_lasso__source-default",
        ],
    },
    {
        "group_id": "cleaned-array-tools",
        "group_label": "Cleaned Array Tools",
        "source_ids": [
            "methylseg_hm450k__cleaned",
            "dnmtools_array__source-default",
        ],
    },
    {
        "group_id": "methylseg-wgbs-raw-vs-cleaned",
        "group_label": "MethylSeg WGBS Raw vs Cleaned",
        "source_ids": [
            "methylseg__raw",
            "methylseg__cleaned",
        ],
    },
    {
        "group_id": "methylseg-array-raw-vs-cleaned",
        "group_label": "MethylSeg HM450K Raw vs Cleaned",
        "source_ids": [
            "methylseg_hm450k__raw",
            "methylseg_hm450k__cleaned",
        ],
    },
]
DISPLAY_GROUP_BY_ID = {group["group_id"]: group for group in DISPLAY_GROUP_SPECS}
PROFILE_COLORS_BY_LABEL = CHROMATIN_PROFILE_COLORS.copy()
EXPECTED_ENV_BIN = Path(
    "/uufs/chpc.utah.edu/common/home/clementm-group1/conda/mambaforge/env/jt_wgbs_analysis/bin"
)
```

```python
def ensure_deeptools_commands() -> tuple[str, str]:
    plot_profile = shutil.which("plotProfile")
    plot_heatmap = shutil.which("plotHeatmap")
    missing = []
    if plot_profile is None:
        missing.append("plotProfile")
    if plot_heatmap is None:
        missing.append("plotHeatmap")
    if missing:
        expected_msg = ""
        if EXPECTED_ENV_BIN.exists():
            expected_msg = (
                f" Expected them on PATH from jt_wgbs_analysis, for example under {EXPECTED_ENV_BIN}."
            )
        raise RuntimeError(
            "Missing deepTools command(s): " + ", ".join(missing) + "."
            + " Start the notebook from the jt_wgbs_analysis environment."
            + expected_msg
        )
    return plot_profile, plot_heatmap


def load_deeptools_outputs_df() -> pd.DataFrame:
    outputs_df = pd.read_csv(DEEPTOOLS_OUTPUTS_PATH, sep="\t")
    outputs_df = outputs_df.sort_values(["sample", "source_order", "source_id"]).reset_index(drop=True)
    return outputs_df


def load_region_manifest_df() -> pd.DataFrame:
    region_df = pd.read_csv(DEEPTOOLS_REGION_MANIFEST_PATH, sep="\t")
    region_df = region_df.sort_values(["sample", "source_order", "source_id"]).reset_index(drop=True)
    return region_df


def select_output_rows_for_sample(outputs_df: pd.DataFrame, sample: str) -> pd.DataFrame:
    sample_rows = outputs_df.loc[outputs_df["sample"].eq(sample)].copy()
    if sample_rows.empty:
        raise KeyError(f"Sample not present in deepTools outputs manifest: {sample}")
    sample_rows = sample_rows.sort_values(["source_order", "source_id"]).reset_index(drop=True)
    return sample_rows


def resolve_saved_matrix_path(outputs_row: pd.Series) -> Path:
    matrix_path = Path(outputs_row["matrix_path"]).expanduser()
    if matrix_path.exists():
        return matrix_path

    sample = str(outputs_row["sample"])
    source_id = str(outputs_row["source_id"])
    fallback_path = CHROMATIN_RESULTS_DIR / "deeptools" / sample / source_id / matrix_path.name
    if fallback_path.exists():
        return fallback_path

    raise FileNotFoundError(
        f"Saved deepTools matrix not found for {sample} {source_id}. Tried {matrix_path} and {fallback_path}."
    )


def get_region_row_for_source(region_df: pd.DataFrame, sample: str, source_id: str) -> pd.Series:
    source_rows = region_df.loc[
        region_df["sample"].eq(sample) & region_df["source_id"].eq(source_id)
    ].copy()
    if source_rows.empty:
        raise KeyError(f"No deepTools region manifest row found for {sample} {source_id}")
    if len(source_rows) != 1:
        raise ValueError(f"Expected one deepTools region row for {sample} {source_id}, found {len(source_rows)}")
    return source_rows.iloc[0]


def get_profile_color_for_label(region_label: str) -> str:
    return PROFILE_COLORS_BY_LABEL.get(region_label, TOOL_HIGHLIGHT_COLORS["Comparator"])


def run_command(command: list[str]) -> None:
    print(" ".join(command))
    result = subprocess.run(command, capture_output=True, text=True)
    if result.returncode != 0:
        if result.stdout:
            print(result.stdout)
        if result.stderr:
            print(result.stderr)
        raise RuntimeError(f"Command failed with exit code {result.returncode}: {' '.join(command)}")
    if result.stdout:
        print(result.stdout)
    if result.stderr:
        print(result.stderr)


def output_paths_for_source(sample: str, source_id: str) -> tuple[Path, Path]:
    profile_out = OUT_DIR / f"{sample}.{source_id}.h3k36me2.profile.png"
    heatmap_out = OUT_DIR / f"{sample}.{source_id}.h3k36me2.heatmap.png"
    return profile_out, heatmap_out


def regenerate_source_plot(
    outputs_row: pd.Series,
    region_row: pd.Series,
    plot_profile_cmd: str,
    plot_heatmap_cmd: str,
) -> dict:
    sample = str(outputs_row["sample"])
    sample_id = str(outputs_row["sample_id"])
    source_id = str(outputs_row["source_id"])
    tool_variant_label = str(outputs_row["tool_variant_label"])
    matrix_path = resolve_saved_matrix_path(outputs_row)
    profile_color = get_profile_color_for_label(tool_variant_label)
    profile_out, heatmap_out = output_paths_for_source(sample, source_id)
    profile_title = str(outputs_row.get("profile_title", f"{sample} H3K36me2 profile - {tool_variant_label}"))
    heatmap_title = str(outputs_row.get("heatmap_title", f"{sample} H3K36me2 heatmap - {tool_variant_label}"))

    profile_command = [
        plot_profile_cmd,
        "--numPlotsPerRow",
        "1",
        "-m",
        str(matrix_path),
        "--perGroup",
        "--colors",
        profile_color,
        "--samplesLabel",
        sample_id,
        "--regionsLabel",
        tool_variant_label,
        "--startLabel",
        "Region start",
        "--endLabel",
        "Region end",
        "--plotTitle",
        profile_title,
        "-out",
        str(profile_out),
        "--plotWidth",
        "14",
        "--plotHeight",
        "10",
    ]

    heatmap_command = [
        plot_heatmap_cmd,
        "-m",
        str(matrix_path),
        "--samplesLabel",
        sample_id,
        "--regionsLabel",
        tool_variant_label,
        "--startLabel",
        "Region start",
        "--endLabel",
        "Region end",
        "--plotTitle",
        heatmap_title,
        "--sortRegions",
        "keep",
        "--heatmapWidth",
        "14",
        "--heatmapHeight",
        "12",
        "--whatToShow",
        "heatmap and colorbar",
        "-out",
        str(heatmap_out),
    ]

    run_command(profile_command)
    run_command(heatmap_command)

    if not profile_out.exists():
        raise FileNotFoundError(f"plotProfile did not create expected file: {profile_out}")
    if not heatmap_out.exists():
        raise FileNotFoundError(f"plotHeatmap did not create expected file: {heatmap_out}")

    return {
        "sample": sample,
        "sample_id": sample_id,
        "source_id": source_id,
        "tool_variant_label": tool_variant_label,
        "matrix_path": str(matrix_path),
        "profile_png": str(profile_out),
        "heatmap_png": str(heatmap_out),
        "visualized_regions": int(region_row["visualized_regions"]),
    }


def display_source_plot(plot_info: dict) -> None:
    display(Markdown(f"## {plot_info['tool_variant_label']}"))
    display(Markdown(f"Matrix: `{plot_info['matrix_path']}`"))
    display(Image(filename=plot_info["profile_png"]))
    display(Image(filename=plot_info["heatmap_png"]))
```

```python
plot_profile_cmd, plot_heatmap_cmd = ensure_deeptools_commands()
deeptools_outputs_df = load_deeptools_outputs_df()
deeptools_region_df = load_region_manifest_df()

if SELECTED_SAMPLES is None:
    samples_to_plot = deeptools_outputs_df["sample"].drop_duplicates().astype(str).tolist()
else:
    samples_to_plot = [
        sample for sample in SELECTED_SAMPLES if sample in set(deeptools_outputs_df["sample"].astype(str))
    ]

if SELECTED_GROUPS is None:
    group_specs_to_plot = list(DISPLAY_GROUP_SPECS)
else:
    group_specs_to_plot = [
        group for group in DISPLAY_GROUP_SPECS if group["group_id"] in set(SELECTED_GROUPS)
    ]

group_source_ids = []
for group_spec in group_specs_to_plot:
    for source_id in group_spec["source_ids"]:
        if source_id not in group_source_ids:
            group_source_ids.append(source_id)

if SELECTED_SOURCE_IDS is None:
    selected_source_ids = set(group_source_ids)
else:
    selected_source_ids = set(group_source_ids).intersection(set(SELECTED_SOURCE_IDS))

filtered_outputs_df = deeptools_outputs_df.loc[
    deeptools_outputs_df["source_id"].isin(selected_source_ids)
].reset_index(drop=True)
filtered_region_df = deeptools_region_df.loc[
    deeptools_region_df["source_id"].isin(selected_source_ids)
].reset_index(drop=True)

display(
    filtered_outputs_df[
        [
            "sample",
            "source_id",
            "tool_variant_label",
            "platform",
            "region_variant",
            "matrix_path",
            "profile_path",
            "heatmap_path",
            "flank_length",
            "region_body_length",
        ]
    ]
)
display(
    filtered_region_df[
        [
            "sample",
            "source_id",
            "tool_variant_label",
            "visualized_regions",
            "source_region_path",
        ]
    ]
)
print("Samples to plot:", samples_to_plot)
print("Groups:", [group["group_id"] for group in group_specs_to_plot])
print("Source ids:", sorted(selected_source_ids))
```

```python
generated_plot_rows = []

for sample in samples_to_plot:
    sample_output_rows = select_output_rows_for_sample(filtered_outputs_df, sample)
    for group_spec in group_specs_to_plot:
        matching_source_rows = []
        for source_id in group_spec["source_ids"]:
            if source_id not in selected_source_ids:
                continue
            source_rows = sample_output_rows.loc[sample_output_rows["source_id"].eq(source_id)].copy()
            if source_rows.empty:
                continue
            outputs_row = source_rows.iloc[0]
            region_row = get_region_row_for_source(filtered_region_df, sample, source_id)
            matching_source_rows.append((outputs_row, region_row))

        if not matching_source_rows:
            continue

        display(Markdown(f"# {sample} - {group_spec['group_label']}"))
        for outputs_row, region_row in matching_source_rows:
            plot_info = regenerate_source_plot(
                outputs_row=outputs_row,
                region_row=region_row,
                plot_profile_cmd=plot_profile_cmd,
                plot_heatmap_cmd=plot_heatmap_cmd,
            )
            generated_plot_rows.append(plot_info)
            display_source_plot(plot_info)

generated_plots_df = pd.DataFrame(generated_plot_rows)
generated_plots_df
```
